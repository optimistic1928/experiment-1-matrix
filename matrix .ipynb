```python
import numpy as np

A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([1.0, 2.0])

print("Matrix A:\n", A)

print("\nTranspose A^T:\n", A.T)

print("\nInverse A^-1:\n", np.linalg.inv(A))

eigvals, eigvecs = np.linalg.eig(A)

print("\nEigenvalues:", np.round(eigvals, 4))

x = np.linalg.solve(A, b)
```
